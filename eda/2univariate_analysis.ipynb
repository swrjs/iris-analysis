```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv('../iris.csv')  
```

TARGET ANALYSIS


```python
df.value_counts('Species')
```

```python
df.value_counts('Species').plot(kind='pie', autopct='%1.2f%%') #kind=bar for bar chart
```

```python
sns.countplot(x='Species', data=df, palette='Set2')
plt.title('Count of Each Species')
```

FEATURE ANALYSIS

```python
df['SepalWidthCm'].max() #min(), mean(), median(), std(), var(), skew(), quantile(0.25), quantile(0.75), describe()
```

```python
plt.figure(figsize=(18, 5))

plt.subplot(1, 4, 1) 
sns.histplot(df['PetalWidthCm'],bins=30, kde=True,color='lightgreen') #kernel density estimation is continuous function approximation
plt.title('Distribution of Petal Width')

plt.subplot(1, 4, 2)
sns.histplot(df['PetalLengthCm'], bins=30, kde=True, color='purple')
plt.title('Distribution of Petal Length')

plt.subplot(1, 4, 3)
sns.histplot(df['SepalLengthCm'], bins=30, kde=True, color='orange')
plt.title('Distribution of Sepal Length')

plt.subplot(1, 4, 4)
sns.histplot(df['SepalWidthCm'], bins=30, kde=True, color='blue')
plt.title('Distribution of Sepal Width')
```

```python
plt.figure(figsize=(18, 5))

plt.subplot(1, 4, 1)
sns.histplot(
    data=df,
    x='PetalWidthCm',
    bins=30,
    kde=True,
    hue='Species',
    palette='Set1',
    element='step',
    alpha=0.35
)
plt.title('Distribution of Petal Width')

plt.subplot(1, 4, 2)
sns.histplot(
    data=df,
    x='PetalLengthCm',
    bins=30,
    kde=True,
    hue='Species',
    palette='Set1',
    element='step',
    alpha=0.35
)
plt.title('Distribution of Petal Length')

plt.subplot(1, 4, 3)
sns.histplot(
    data=df,
    x='SepalLengthCm',
    bins=30,
    kde=True,
    hue='Species',
    palette='Set1',
    element='step',
    alpha=0.35
)
plt.title('Distribution of Sepal Length')

plt.subplot(1, 4, 4)
sns.histplot(
    data=df,
    x='SepalWidthCm',
    bins=30,
    kde=True,
    hue='Species',
    palette='Set1',
    element='step',
    alpha=0.35
)
plt.title('Distribution of Sepal Width')

plt.tight_layout()
plt.show()
```

The Second set of histogram shows that Sepal Width and Sepal Length overlap the most hence the least unique parameter of each species
