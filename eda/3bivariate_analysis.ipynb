```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

a=pd.read_csv('../iris.csv')
```

```python
a=a.iloc[:, 1:] 
```

```python
sns.pairplot(a, hue='Species', palette='husl')
plt.show()
```

By looking at the graphs
PetalWidth and PetalLength looks like the most distinctive feature of each species while
SepalWidth and SepalLength seem to overalap a lot

```python
plt.figure(figsize=(8, 6))
correlation_matrix = a.drop('Species', axis=1).corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Feature Correlation Heatmap')
```

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(x='PetalLengthCm', y='PetalWidthCm', hue='Species', data=a, palette='Set1')
plt.title('Petal Length vs Petal Width')
plt.show()
```
